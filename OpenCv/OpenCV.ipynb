# **Contour**

```
image_path = "machine_learning/OpenCv/images/basic_shapes.png"
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Could not load image from {image_path}. Please check if the path is correct and the file exists.")
else:
    print("Original Image:")
    cv2_imshow(image) # Display original image first

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    print("Grayscale Image:")
    cv2_imshow(gray) # Display grayscale image
```

```
# Apply binary inverse thresholding to create white shapes on a black background
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)
print("Binary Inverse Thresholded Image:")
cv2_imshow(thresh)
```

```
non_zero_pixels = cv2.countNonZero(thresh)
print(f"Number of non-zero pixels in thresh image: {non_zero_pixels}")

if non_zero_pixels == 0:
    print("The thresh image is completely black (empty).")
else:
    print("The thresh image contains non-zero pixels.")
```

**Internal contour**

```
cnts = cv2.findContours(thresh.copy(), cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
cnts = imutils.grab_contours(cnts)
clone = image.copy()
cv2.drawContours(clone, cnts, -1, (0, 255, 0), 2)
```

```
cv2_imshow(clone)
```

**External Contour**

```
cnts = cv2.findContours(thresh.copy(), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cnts = imutils.grab_contours(cnts)
cv2.drawContours(clone, cnts, -1, (0, 255, 0), 2)
```

```
for c in cnts:
	# construct a mask by drawing only the current contour
	mask = np.zeros(gray.shape, dtype="uint8")
	cv2.drawContours(mask, [c], -1, 255, -1)

	# show the images
	#cv2.imshow("Image", image)
	#cv2.imshow("Mask", mask)
	cv2_imshow(cv2.bitwise_and(image, image, mask=mask))
```

# **License Plate Detection**

```
image = cv2.imread("machine_learning/OpenCv/images/1.jpg")
image = imutils.resize(image, width=500)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```
cv2_imshow(image)
```

```
(T, threshInv) = cv2.threshold(gray, 0, 255,cv2.THRESH_BINARY_INV | cv2.THRESH_OTSU)
cv2_imshow(threshInv)
```

```
cnts = cv2.findContours(threshInv.copy(), cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
cnts = imutils.grab_contours(cnts)
clone = image.copy()

for cnt in cnts:
	x,y,w,h = cv2.boundingRect(cnt)

	if (w<120 and w>90 and h<70 and h>30):
		print (x,y,w,h)
		cv2.rectangle(image,(x,y),(x+w,y+h),(0,255,0),2)
		cv2_imshow(image)
```

```
import numpy as np

(T, threshInv) = cv2.threshold(gray, 0, 255,cv2.THRESH_BINARY_INV | cv2.THRESH_OTSU)
cv2_imshow(threshInv)

kernel = np.ones((3,3),np.uint8)
threshInv = cv2.erode(threshInv,kernel, iterations=3)
cv2_imshow(threshInv)
```

```
threshInv = cv2.dilate(threshInv, kernel, iterations=3)
cv2_imshow(threshInv)
```
